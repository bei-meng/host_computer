```python
getbits = lambda x, y : (x >> y) & 1

def map(chip_sel): 
    gpio = 0
    index_map = {0: 1, 1: 3, 2: 5, 3: 7, 4: 9}
    for key, value in index_map.items():
        a_key = ((chip_sel >> key) & 1)
        gpio |= (a_key << value)
    return gpio

map_chipsel_gpio = [0] * 19
for i in range(1, 19):
    map_chipsel_gpio[i] = map(i)

for i in range(19):
    gpio = map_chipsel_gpio[i]
    print(f'i: {i}, gpio: {getbits(gpio, 9)} {getbits(gpio, 7)} {getbits(gpio, 5)} {getbits(gpio, 3)} {getbits(gpio, 1)}')
```
